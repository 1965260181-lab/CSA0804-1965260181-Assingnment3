```python
# Creating bus stops using tuples

bus_stops = (
    ("Stop 1", 10),
    ("Stop 2", 20),
    ("Stop 3", 30),
    ("Stop 4", 40)
)

print("Bus Stop Details:")
for stop in bus_stops:
    print(stop)
```

```python
bus_map = {
    (0, 0): "Central Bus Stop",
    (0, 1): "Railway Station",
    (1, 0): "Airport",
    (1, 1): "College Bus Stop"
}

print("Bus Stop Map:")

for position, stop in bus_map.items():
    print(position, ":", stop)
```

```python
rows = 3
columns = 3

bus_grid = [[0 for j in range(columns)] for i in range(rows)]

bus_grid[0][0] = "Central"
bus_grid[0][1] = "Railway"
bus_grid[1][0] = "Airport"
bus_grid[1][1] = "College"

for row in bus_grid:
    print(row)
```

```python
import copy

original = [[10, 20], [30, 40]]

alias = original
shallow_copy = original.copy()
deep_copy = copy.deepcopy(original)

original[0][0] = 100

print("Original:", original)
print("Alias:", alias)
print("Shallow Copy:", shallow_copy)
print("Deep Copy:", deep_copy)
```

```python
bus_stop = ("Central", 12, "Chennai")

name, number, city = bus_stop

print("Bus Stop Name:", name)
print("Bus Stop Number:", number)
print("City:", city)
```

```python
bus_data = ("Central", ["Stop A", "Stop B"])

bus_data[1].append("Stop C")

print(bus_data)
```

```python
import sys

my_list = list(range(1000000))
my_tuple = tuple(range(1000000))

list_size = sys.getsizeof(my_list)
tuple_size = sys.getsizeof(my_tuple)

print("List Memory:", list_size, "bytes")
print("Tuple Memory:", tuple_size, "bytes")
```

### Design and Analysis

**1. Tuples**

Tuples are used to store bus-stop details because they are ordered and immutable.

**2. Dictionary**

A dictionary stores bus-stop locations using tuple keys such as `(0, 0)`. Tuples are suitable dictionary keys because they are immutable.

**3. List Comprehension**

List comprehension is used to create a 2D grid easily and avoid the list aliasing problem.

**4. Aliasing**

Aliasing occurs when two variables refer to the same object. Changes made through one variable affect the other.

**5. Shallow Copy**

A shallow copy creates a new outer list but shares the nested objects.

**6. Deep Copy**

A deep copy creates a completely independent copy, including nested objects.

**7. Tuple Unpacking**

Tuple unpacking assigns tuple values to separate variables.

**8. Memory Comparison**

The `sys.getsizeof()` function is used to compare the memory occupied by lists and tuples.

### Time Complexity Analysis

| Operation | Time Complexity |
|---|---|
| Tuple element access | O(1) |
| Dictionary key access | O(1) average |
| Dictionary insertion | O(1) average |
| List element access | O(1) |
| List insertion | O(n) |
| Shallow copy | O(n) |
| Deep copy | O(n + d) |
| Creating a 2D grid | O(r × c) |

Here, `n` represents the number of elements, `r` represents rows, `c` represents columns and `d` represents the total nested elements.

### Comparison

- Tuples use less memory than lists in many cases.
- Dictionary lookup using tuple keys is efficient.
- Deep copy takes more time than shallow copy because it copies nested objects.
- List comprehension creates independent rows and avoids aliasing problems.

### Conclusion

The City Bus-Stop Map program demonstrates the use of tuples, dictionaries and nested lists in Python.

Tuples store fixed bus-stop information, while dictionaries provide efficient access using tuple keys.

Shallow copy and deep copy demonstrate the differences in copying nested data. List comprehension helps create independent rows and prevents aliasing problems.
