```python
import time

# Railway timetable using nested dictionaries
trains = {
    "101": {
        "name": "Chennai Express",
        "stations": [
            ("Chennai", "06:00", "06:10"),
            ("Vellore", "08:00", "08:05"),
            ("Bangalore", "11:00", "11:15")
        ],
        "running_days": {"Monday", "Wednesday", "Friday"}
    },

    "102": {
        "name": "Madurai Express",
        "stations": [
            ("Chennai", "07:00", "07:10"),
            ("Trichy", "10:00", "10:05"),
            ("Madurai", "12:00", "12:15")
        ],
        "running_days": {"Tuesday", "Thursday", "Saturday"}
    },

    "103": {
        "name": "Coimbatore Express",
        "stations": [
            ("Chennai", "08:00", "08:10"),
            ("Salem", "11:00", "11:05"),
            ("Coimbatore", "13:00", "13:15")
        ],
        "running_days": {"Monday", "Tuesday", "Friday"}
    },

    "104": {
        "name": "Trichy Express",
        "stations": [
            ("Chennai", "09:00", "09:10"),
            ("Villupuram", "11:00", "11:05"),
            ("Trichy", "13:00", "13:15")
        ],
        "running_days": {"Wednesday", "Thursday", "Sunday"}
    },

    "105": {
        "name": "Bangalore Express",
        "stations": [
            ("Chennai", "10:00", "10:10"),
            ("Vellore", "12:00", "12:05"),
            ("Bangalore", "15:00", "15:15")
        ],
        "running_days": {"Monday", "Saturday", "Sunday"}
    }
}

# Display complete timetable
print("RAILWAY TIMETABLE")

for train_no, details in trains.items():
    print("\nTrain Number:", train_no)
    print("Train Name:", details["name"])
    print("Running Days:", details["running_days"])

    for station in details["stations"]:
        print(station)

# Find trains stopping at a particular station on a given day
search_station = "Vellore"
search_day = "Monday"

results = []

for train_no, details in trains.items():
    if search_day in details["running_days"]:
        for station in details["stations"]:
            if station[0] == search_station:
                results.append((train_no, details["name"], station[2]))

# Sort by departure time
results.sort(key=lambda x: x[2])

print("\nTrains stopping at", search_station, "on", search_day)

if results:
    for train in results:
        print("Train:", train[0], "| Name:", train[1],
              "| Departure:", train[2])
else:
    print("No trains found")

# Access departure time of third station of train 101
print("\nThird Station Departure Time:",
      trains["101"]["stations"][2][2])

# Station index for faster searching
station_index = {}

for train_no, details in trains.items():
    for station in details["stations"]:
        station_index.setdefault(station[0], []).append(train_no)

print("\nStation Index:")
print(station_index)

# Compare searching without and with index
start = time.time()

for train_no, details in trains.items():
    for station in details["stations"]:
        if station[0] == "Vellore":
            pass

without_index = time.time() - start

start = time.time()
indexed_trains = station_index.get("Vellore", [])
with_index = time.time() - start

print("\nSearch Without Index:", without_index)
print("Search With Index:", with_index)

# Performance testing with different train counts
for size in [100, 1000, 10000]:
    test_trains = {}

    for i in range(size):
        test_trains[str(i)] = {
            "stations": [
                ("Chennai", "06:00", "06:10"),
                ("Vellore", "08:00", "08:05"),
                ("Bangalore", "11:00", "11:15")
            ],
            "running_days": {"Monday", "Wednesday"}
        }

    start = time.time()

    for train_no, details in test_trains.items():
        for station in details["stations"]:
            if station[0] == "Vellore":
                pass

    scan_time = time.time() - start

    start = time.time()

    index = {}
    for train_no, details in test_trains.items():
        for station in details["stations"]:
            index.setdefault(station[0], []).append(train_no)

    index_time = time.time() - start

    print("\nNumber of Trains:", size)
    print("Without Index:", scan_time)
    print("Building Index:", index_time)

# Unknown station test
print("\nUnknown Station Test:", station_index.get("Mumbai", []))
```

The Railway Timetable program uses nested data structures to store and manage train information.

1. Dictionary: Stores train details using train numbers as keys.
2. Nested Dictionary: Stores train names, stations and running days.
3. List: Stores multiple station details.
4. Tuple: Stores station name, arrival time and departure time.
5. Set: Stores the days on which each train operates.
6. Sorting: Arranges matching trains according to departure time.
7. Station Index: Helps locate trains stopping at a particular station.

Expression to access the departure time of the third station of train 101:

trains["101"]["stations"][2][2]

This program makes railway information easier to store, search and display.

4. Time Complexity Analysis
Operation	Time Complexity
Dictionary lookup	O(1) average
Set membership	O(1) average
Station search	O(T × S)
Sorting results	O(R log R)
Creating station index	O(T × S)
Indexed station lookup	O(1) average to retrieve the list


Here, T = number of trains, S = stations per train, and R = matching trains.

The Railway Timetable program demonstrates the use of nested dictionaries, lists, tuples and sets in Python. It stores train details, identifies trains stopping at a particular station on a given day and sorts them by departure time.
