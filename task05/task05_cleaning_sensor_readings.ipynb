```python
import time
import sys

# Temperature sensor readings in Celsius
readings = [10, 25, 55, -60, 5, 3, 8, 45, 60, -10, 2, 7, 15]

# Remove invalid readings (below -50 or above 50)
valid_readings = [x for x in readings if -50 <= x <= 50]

print("Original Readings:", readings)
print("Valid Readings:", valid_readings)

# Convert Celsius to Fahrenheit
fahrenheit = [(x * 9/5) + 32 for x in valid_readings]

print("\nFahrenheit Readings:", fahrenheit)

# Find alarm minutes (temperature outside 2 to 8°C)
alarm_minutes = [i + 1 for i, x in enumerate(valid_readings) if x < 2 or x > 8]

print("\nAlarm Minutes:", alarm_minutes)

# Using for loop
loop_result = []
for x in readings:
    if -50 <= x <= 50:
        loop_result.append(x)

print("\nFor Loop Result:", loop_result)

# Using filter and lambda
filter_result = list(filter(lambda x: -50 <= x <= 50, readings))

print("Filter Result:", filter_result)

# Using map and lambda
map_result = list(map(lambda x: (x * 9/5) + 32, valid_readings))

print("Map Result:", map_result)

# Verify all methods produce the same valid readings
print("\nAll Results Equal:", valid_readings == loop_result == filter_result)

# Performance testing
for size in [10000, 100000, 1000000]:
    data = [i % 100 - 50 for i in range(size)]

    start = time.time()
    result = [x for x in data if -50 <= x <= 50]
    end = time.time()

    print("\nReadings:", size)
    print("List Comprehension Time:", end - start)
    print("Memory Used:", sys.getsizeof(result), "bytes")

# Generator expression alternative
generator_result = (x for x in readings if -50 <= x <= 50)

print("\nGenerator Result:", list(generator_result))
```

3. Design and Analysis
- List comprehension: Creates a new list using a single expression.
- Filtering: Removes invalid sensor readings.
- Map: Converts Celsius into Fahrenheit.
- Filter: Selects readings within the valid range.
- Lambda: Creates a small anonymous function.
- Enumerate: Provides the index and value of each reading.
- Generator expression: Produces values one at a time, reducing memory usage.
