```python
import re
import time
import random

# Contact records from three sources
phone_contacts = [
    {"name": "Kaviya V", "phone": "98765-43210", "email": "kaviya@gmail.com"},
    {"name": "Keerthana P", "phone": "91234 56789", "email": "keerthana@gmail.com"},
    {"name": "Ruksana K", "phone": "9876543211", "email": "ruksana@gmail.com"}
]

old_phone_contacts = [
    {"name": "KAVIYA V", "phone": "+91 98765 43210", "email": "KAVIYA@gmail.com"},
    {"name": "Keerthana", "phone": "9123456789", "email": ""},
    {"name": "Priya", "phone": "9000011111", "email": "priya@gmail.com"}
]

sim_contacts = [
    {"name": "Kaviya", "phone": "9876543210", "email": ""},
    {"name": "Ruksana K", "phone": "", "email": "ruksana@gmail.com"},
    {"name": "Arun", "phone": "9000022222", "email": "arun@gmail.com"}
]

# Normalize phone number
def normalize_phone(phone):
    return re.sub(r'\D', '', phone)

# Normalize name
def normalize_name(name):
    return " ".join(name.lower().split())

# Normalize email
def normalize_email(email):
    return email.strip().lower()

# Store merged contacts
merged_contacts = []

# Indexes for searching
phone_index = {}
email_index = {}

# Merge contacts
all_contacts = phone_contacts + old_phone_contacts + sim_contacts

for contact in all_contacts:

    name = normalize_name(contact.get("name", ""))
    phone = normalize_phone(contact.get("phone", ""))
    email = normalize_email(contact.get("email", ""))

    matches = set()

    if phone and phone in phone_index:
        matches.add(phone_index[phone])

    if email and email in email_index:
        matches.add(email_index[email])

    if matches:
        main_id = min(matches)

        for other_id in matches:
            if other_id != main_id:
                main = merged_contacts[main_id]
                other = merged_contacts[other_id]

                main["names"].update(other["names"])
                main["phones"].update(other["phones"])
                main["emails"].update(other["emails"])

                for p in other["phones"]:
                    phone_index[p] = main_id

                for e in other["emails"]:
                    email_index[e] = main_id

                other["active"] = False

        merged_contacts[main_id]["names"].add(name)

        if phone:
            merged_contacts[main_id]["phones"].add(phone)

        if email:
            merged_contacts[main_id]["emails"].add(email)

    else:
        new_id = len(merged_contacts)

        new_contact = {
            "names": {name} if name else set(),
            "phones": {phone} if phone else set(),
            "emails": {email} if email else set(),
            "active": True
        }

        merged_contacts.append(new_contact)

        if phone:
            phone_index[phone] = new_id

        if email:
            email_index[email] = new_id

    if phone:
        phone_index[phone] = min(matches) if matches else new_id

    if email:
        email_index[email] = min(matches) if matches else new_id

# Display final merged contacts
print("MERGED CONTACTS")

count = 1

for contact in merged_contacts:
    if contact["active"]:
        print("\nContact", count)
        print("Names:", contact["names"])
        print("Phone Numbers:", contact["phones"])
        print("Emails:", contact["emails"])
        count += 1

print("\nTotal Unique Contacts:", count - 1)

# Test missing fields
test_contact = {"name": "Unknown", "phone": "", "email": ""}
print("\nMissing Fields Test:", test_contact)

# Performance testing
for size in [1000, 100000]:

    test_data = []

    for i in range(size):
        test_data.append({
            "name": "Person " + str(i),
            "phone": str(9000000000 + i),
            "email": "person" + str(i) + "@gmail.com"
        })

    start = time.time()

    test_index = {}

    for contact in test_data:
        phone = normalize_phone(contact["phone"])
        test_index[phone] = contact

    end = time.time()

    print("\nNumber of Contacts:", size)
    print("Dictionary Processing Time:", end - start)

# Verify normalization
print("\nPhone Normalization:")
print(normalize_phone("+91 98765-43210"))

print("\nName Normalization:")
print(normalize_name("  KAVIYA V  "))

print("\nEmail Normalization:")
print(normalize_email(" KAVIYA@GMAIL.COM "))
```

1. Dictionary: Stores and searches contacts efficiently.
2. Sets: Store unique names, phone numbers and email addresses.
3. Phone Normalization: Removes spaces and special characters from phone numbers.
4. Name Normalization: Converts names into lowercase and removes extra spaces.
5. Email Normalization: Converts email addresses into lowercase.
6. Duplicate Detection: Uses phone numbers and email addresses to identify matching contacts.
7. Contact Merging: Combines information belonging to the same contact.
8. Missing Fields: Allows contacts with missing phone numbers or email addresses.

The program reduces duplicate contact entries and maintains organized contact information.

4. Time Complexity Analysis
Operation	Time Complexity
Phone normalization	O(p)
Name normalization	O(n)
Dictionary lookup	O(1) average
Set insertion	O(1) average
Contact merging	O(k)
Processing all contacts	O(N) average


Here, N represents the total number of contacts, and k represents the number of matching contacts being merged.
