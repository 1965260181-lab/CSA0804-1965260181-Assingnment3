```python
def calculate_average(data):
    averages = []

    for i in range(len(data) - 6):
        average = sum(data[i:i+7]) / 7
        averages.append(average)

    return averages


print("Empty List:", calculate_average([]))
print("Small List:", calculate_average([10, 20, 30]))
```

```python
import time
import random

def naive_method(data):
    result = []
    for i in range(len(data) - 6):
        result.append(sum(data[i:i+7]) / 7)
    return result


def running_sum_method(data):
    result = []
    if len(data) < 7:
        return result

    current_sum = sum(data[:7])
    result.append(current_sum / 7)

    for i in range(7, len(data)):
        current_sum += data[i] - data[i-7]
        result.append(current_sum / 7)

    return result


for size in [365, 36500, 365000]:
    data = [random.randint(0, 50) for i in range(size)]

    start = time.time()
    naive_method(data)
    naive_time = time.time() - start

    start = time.time()
    running_sum_method(data)
    running_time = time.time() - start

    print("Data Size:", size)
    print("Naive Time:", naive_time)
    print("Running Sum Time:", running_time)
    print()
```

```python
data = [10, 20, 0, 15, 30, 25, 5, 12, 18]

naive_result = naive_method(data)
running_result = running_sum_method(data)

print("Naive Result:", naive_result)
print("Running Sum Result:", running_result)
print("Both Results Equal:", naive_result == running_result)
```

### Design and Analysis

**1. Data Structure Used: List**

A list is used to store 365 daily rainfall readings. Lists are suitable because they can store multiple values and allow easy modification.

**2. List Slicing**

List slicing is used to access specific rainfall records.

- `readings[:7]` – First 7 days.
- `readings[7:14]` – Second week.
- `readings[-7:]` – Last 7 days.

**3. List Operations**

- `append()` – Adds a new rainfall reading.
- `insert()` – Inserts a reading at a particular position.
- `pop()` – Removes an element.
- `sort()` – Arranges readings in ascending order.

**4. Moving Average**

The 7-day moving average calculates the average rainfall for every consecutive 7-day period.

**5. Dry Days**

Days with zero rainfall are identified and stored in a separate list.

### Time Complexity Analysis

| Operation | Time Complexity |
|---|---|
| Accessing a list element | O(1) |
| Appending an element | O(1) average |
| Inserting an element | O(n) |
| Removing an element | O(n) |
| Sorting a list | O(n log n) |
| Finding maximum and minimum | O(n) |
| Naive moving average | O(n × 7), equivalent to O(n) for a fixed 7-day window |
| Running sum moving average | O(n) |

**Comparison:**

- Naive Method calculates the sum of 7 values for every window.
- Running Sum Method updates the previous sum by adding the new value and removing the oldest value.
- Running Sum Method performs fewer repeated calculations and is more efficient.

### Conclusion

The Daily Rainfall Records program successfully stores and analyses 365 days of rainfall data using Python lists and slicing.

It calculates weekly rainfall totals, maximum and minimum rainfall, 7-day moving averages and identifies dry days.

The Running Sum Method is more efficient than the Naive Method because it avoids repeated calculations.

### Alternative Methods

**1. Deque**

A deque with `maxlen=7` stores the latest seven rainfall readings. Older readings are automatically removed when new readings are added.

**2. Dictionary**

A dictionary can store rainfall readings using dates as keys and rainfall values as values. It helps access rainfall data for a particular date easily.
