```python
import random
import time

# Student elective registrations
student1 = {"Python", "AI", "Cyber Security", "Data Science"}
student2 = {"Python", "Cloud Computing", "AI", "Networking"}
student3 = {"Cyber Security", "AI", "Networking"}

# Union
all_electives = student1 | student2 | student3
print("All Electives:", sorted(all_electives))

# Intersection
common_electives = student1 & student2
print("\nCommon Electives:", sorted(common_electives))

# Difference
difference = student1 - student2
print("\nStudent 1 Only:", sorted(difference))

# Symmetric Difference
symmetric_difference = student1 ^ student2
print("\nDifferent Electives:", sorted(symmetric_difference))

# Remove duplicate registrations
registrations = ["Python", "AI", "Python", "Networking", "AI"]
unique_registrations = set(registrations)
print("\nUnique Registrations:", sorted(unique_registrations))

# Check elective availability
elective = "Python"
print("\nIs Python registered?", elective in student1)

# Find common electives among all students
common_all = student1 & student2 & student3
print("Common to All Students:", sorted(common_all))

# Set membership and list membership performance
for size in [10000, 100000, 1000000]:
    numbers = list(range(size))
    number_set = set(numbers)

    search_value = size - 1

    start = time.time()
    search_value in numbers
    list_time = time.time() - start

    start = time.time()
    search_value in number_set
    set_time = time.time() - start

    print("\nNumber of Elements:", size)
    print("List Search Time:", list_time)
    print("Set Search Time:", set_time)

# Nested loops vs set intersection
list1 = list(range(1000))
list2 = list(range(500, 1500))

start = time.time()
nested_result = []
for a in list1:
    for b in list2:
        if a == b:
            nested_result.append(a)
nested_time = time.time() - start

start = time.time()
set_result = set(list1) & set(list2)
set_time = time.time() - start

print("\nNested Loop Time:", nested_time)
print("Set Intersection Time:", set_time)
print("Common Values:", sorted(set_result))
```

3. Design and Analysis
- Set: Stores unique elective registrations.
- Union (|): Combines all electives.
- Intersection (&): Finds common electives.
- Difference (-): Finds electives present in one set but not another.
- Symmetric Difference (^): Finds elements present in either set, but not both.
- Membership (in): Checks whether an elective exists.
- Sets automatically remove duplicate values.

4. Time Complexity Analysis
Operation	Time Complexity
Set insertion	O(1) average
Set membership	O(1) average
Union	O(n + m)
Intersection	O(min(n, m)) average
Difference	O(n) average
List membership	O(n)
Nested loops	O(n × m)
